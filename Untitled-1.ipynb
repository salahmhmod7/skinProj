```python
# -------------------------------------------------
# Cell 8: Basic image metrics
# -------------------------------------------------
def image_metrics(img_path):
    """Return resolution, aspect ratio, and channel count."""
    with Image.open(img_path) as img:
        width, height = img.size
        resolution = f"{width}x{height}"
        aspect = width / height
        channels = len(img.getbands())  # 1 for grayscale, 3 for RGB
    return resolution, aspect, channels

# Sample a few images from each class to compute metrics
sample_paths = []
for cls in df['dx'].unique()[:3]:  # first 3 classes for brevity
    row = df[df['dx'] == cls].iloc[0]
    p = row['image_path']
    if p:
        sample_paths.append(p)

metrics = [image_metrics(p) for p in sample_paths]
metrics_df = pd.DataFrame(metrics, columns=['Resolution', 'Aspect_Ratio', 'Channels'])
print(metrics_df)
```

```python
# -------------------------------------------------
# Cell 3: Impute missing age using median age per dx group
# -------------------------------------------------
# Compute median age for each diagnostic class
median_age_by_dx = df.groupby('dx')['age'].median()

# Fill NaNs; if a class has no non‑NaN ages (unlikely) fall back to overall median
overall_median = df['age'].median()

# Fill NaNs using the mapped median by dx category
df['age'] = df['age'].fillna(df['dx'].map(median_age_by_dx))

# Fallback for any still‑NaN (e.g., a class that was completely missing)
df['age'] = df['age'].fillna(overall_median)

print("\n=== Missing age after imputation ===")
print(df['age'].isnull().sum())
```

```python
# -------------------------------------------------
# Cell 1: Imports & global settings
# -------------------------------------------------
# %matplotlib inline   # uncomment if you run the notebook with the magic

import os
import glob
import math
import random

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

# --- Deep‑learning libraries -------------------------------------------------
import torch
from torch.utils.data import Dataset, DataLoader  # Dataset & DataLoader
from torchvision import transforms as T_vision          # optional fallback

# --- albumentations ---------------------------------------------------------
import albumentations as A
from albumentations import pytorch as AT               # To convert → torch Tensor

# --- sklearn ---------------------------------------------------------------
from sklearn.utils.class_weight import compute_class_weight

# -------------------------------------------------------------------------
# Basic style / reproducibility
# -------------------------------------------------------------------------
sns.set(style="whitegrid", font_scale=1.2)
plt.rcParams.update({
    "figure.figsize": (10, 6),
    "font.size": 12,
})

# Random seed (optional, helps reproducibility)
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)
```

```python
# -------------------------------------------------
# Cell 2: Load train / val / test metadata
# -------------------------------------------------
data_dir = "data"                     # relative path from notebook root

train_df = pd.read_csv(os.path.join(data_dir, "train_metadata.csv"))
val_df   = pd.read_csv(os.path.join(data_dir, "val_metadata.csv"))
test_df  = pd.read_csv(os.path.join(data_dir, "test_metadata.csv"))

print("Train rows :", train_df.shape[0])
print("Val   rows :", val_df.shape[0])
print("Test  rows :", test_df.shape[0])
# Expected: ~8012 / ~1002 / ~1001 (exact numbers depend on the split)
```

```python
# -------------------------------------------------
# Cell 3: Compute balanced class weights (sklearn) → PyTorch tensor
# -------------------------------------------------
# The 'dx' column contains the short codes (nv, mel, …)
train_labels = train_df['dx'].values          # numpy array of shape (N,)

# sklearn expects integer labels → map them to 0 … 6
dx_to_idx = {code: i for i, code in enumerate(sorted(train_df['dx'].unique()))}
train_idx = np.array([dx_to_idx[l] for l in train_labels])

# compute_class_weight returns weights for each class
class_weights_np = compute_class_weight(
    class_weight='balanced',
    classes=np.unique(train_idx),
    y=train_idx
)

# Convert to a PyTorch FloatTensor (will be used inside CrossEntropyLoss)
class_weights_torch = torch.tensor(class_weights_np, dtype=torch.float32)

print("\n=== Class weights (balanced) ===")
for code, w in zip(sorted(dx_to_idx.keys()), class_weights_np):
    print(f"{code:5s} : {w:.4f}")

print("\nTensor shape:", class_weights_torch.shape)
print("Device:", class_weights_torch.device)   # will be moved to GPU later if needed
```

```python
import albumentations as A
from albumentations.pytorch import ToTensorV2

# -------------------------------------------------
# Cell 4: albumentations train / val / test transforms
# -------------------------------------------------
# Normalisation constants (ImageNet mean / std)
IMG_MEAN = [0.485, 0.456, 0.406]      # R, G, B
IMG_STD  = [0.229, 0.224, 0.225]

# -----------------------------------------------------------------
# Helper: common resizing + normalization step (used by val/test)
# -----------------------------------------------------------------
def get_base_transform():
    """Resize + Normalize (no augmentation)."""
    return A.Compose([
        A.Resize(224, 224), 
        A.Normalize(mean=IMG_MEAN, std=IMG_STD, max_pixel_value=255.0),
        # استخدام ToTensorV2 بدلاً من AT.ToTensor
        ToTensorV2(),
    ])

# -----------------------------------------------------------------
# Train augmentations – heavy but realistic for dermoscopy
# -----------------------------------------------------------------
train_transform = A.Compose([
    A.Resize(224, 224),

    # ----- geometric -------------------------------------------------
    A.HorizontalFlip(p=0.5),
    A.VerticalFlip(p=0.3), 
    A.RandomRotate90(p=0.5),
    # التحديث: استخدام Affine بدلاً من ShiftScaleRotate
    A.Affine(scale=(0.8, 1.2), translate_percent=(-0.1, 0.1), rotate=(-30, 30), p=0.7),

    # ----- colour / brightness ---------------------------------------
    A.ColorJitter(brightness=0.2, contrast=0.2, saturation=0.2, hue=0.1, p=0.5),
    # التحديث: استخدام std_range بدلاً من var_limit
    A.GaussNoise(std_range=(0.1, 0.3), p=0.3),

    # ----- optional mild distortions (kept subtle) ------------------
    # التحديث: إزالة alpha_affine غير المدعومة
    A.ElasticTransform(alpha=1, sigma=5, p=0.3),
    A.GridDistortion(p=0.3),

    # ----- final normalisation + tensor conversion ------------------
    A.Normalize(mean=IMG_MEAN, std=IMG_STD, max_pixel_value=255.0),
    ToTensorV2(),
])

# -----------------------------------------------------------------
# Validation / Test transforms – *only* resize + normalisation
# -----------------------------------------------------------------
val_transform = get_base_transform()
test_transform  = get_base_transform()
```

```python
# -------------------------------------------------
# Cell 5: Custom Dataset (PyTorch) for HAM10000
# -------------------------------------------------
class Ham10000Dataset(Dataset):
    """
    Parameters
    ----------
    df : pandas.DataFrame
        Must contain columns: 'image_id', 'dx' (short code), and optionally
        'age', 'sex', 'localization', etc.
    transform : albumentations Compose object
        The augmentation / normalisation pipeline to apply.
    data_folders : list of str
        Paths relative to the notebook root where the .jpg files live,
        e.g. ["data/HAM10000_images_part_1", "data/HAM10000_images_part_2"].
    """
    def __init__(self, df, transform, data_folders=None):
        self.df = df.reset_index(drop=True)
        self.transform = transform

        # Build a dictionary mapping image_id -> full path.
        # We check both parts; later images override earlier ones (harmless if no overlap).
        self.id_to_path = {}
        if data_folders:
            for folder in data_folders:
                # List all .jpg files; assume filenames are just the image_id with .jpg extension
                for img_file in os.listdir(folder):
                    if img_file.lower().endswith(('.jpg', '.jpeg', '.png')):
                        img_id = img_file.split('.')[0]
                        # Store; later folder wins if duplicate ids exist
                        self.id_to_path[img_id] = os.path.join(folder, img_file)

        # Sanity check: verify that every image_id in the DataFrame has a path
        missing = set(self.df['image_id']) - set(self.id_to_path.keys())
        if missing:
            print(f"⚠️  Warning: {len(missing)} image_ids have no corresponding file on disk.")
        else:
            print(f"✅ All {len(self.df)} image_ids found on disk.")

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        row = self.df.iloc[idx]

        # -----------------------------------------------------------------
        # 1️⃣ Load image
        # -----------------------------------------------------------------
        img_id = row['image_id']
        img_path = self.id_to_path.get(img_id)
        if img_path is None or not os.path.exists(img_path):
            # fallback: return a blank tensor (or raise)
            raise FileNotFoundError(f"Image {img_id} not found at {img_path}")

        image = Image.open(img_path).convert('RGB')   # always RGB
        image = np.array(image)                         # H x W x C, dtype=uint8

        # -----------------------------------------------------------------
        # 2️⃣ Apply albumentations transform
        # -----------------------------------------------------------------
        augmented = self.transform(image=image)
        image = augmented['image']                       # torch Tensor (C, H, W) in [0,1]

        # -----------------------------------------------------------------
        # 3️⃣ Encode target label (dx) → integer index 0‑6
        # -----------------------------------------------------------------
        dx_code = row['dx']
        label_idx = dx_to_idx[dx_code]                 # use the same mapping from Cell 3

        return image, label_idx, img_id               # you can also return extra metadata
```

```python
# -------------------------------------------------
# Cell 6: Create Dataset objects and DataLoaders
# -------------------------------------------------
# Define the three folders that contain the actual image files
IMAGE_FOLDERS = [
    "data/HAM10000_images_part_1",
    "data/HAM10000_images_part_2"
]

# -----------------------------------------------------------------
# Initialise the three datasets (they will internally verify paths)
# -----------------------------------------------------------------
train_dataset = Ham10000Dataset(df=train_df, transform=train_transform,
                               data_folders=IMAGE_FOLDERS)
val_dataset   = Ham10000Dataset(df=val_df,   transform=val_transform,
                               data_folders=IMAGE_FOLDERS)
test_dataset  = Ham10000Dataset(df=test_df,  transform=test_transform,
                               data_folders=IMAGE_FOLDERS)

# -----------------------------------------------------------------
# DataLoaders – shuffle only for training
# -----------------------------------------------------------------
BATCH_SIZE = 16
NUM_WORKERS = 0          # adjust: 0 on Windows if you get multiprocessing errors

train_loader = DataLoader(train_dataset,
                          batch_size=BATCH_SIZE,
                          shuffle=True,
                          num_workers=NUM_WORKERS,
                          pin_memory=True)   # faster GPU transfer

val_loader = DataLoader(val_dataset,
                        batch_size=BATCH_SIZE,
                        shuffle=False,
                        num_workers=NUM_WORKERS,
                        pin_memory=True)

test_loader = DataLoader(test_dataset,
                         batch_size=BATCH_SIZE,
                         shuffle=False,
                         num_workers=NUM_WORKERS,
                         pin_memory=True)

print("\n=== DataLoader summary ===")
print(f"Train batches per epoch: {len(train_loader)}  (≈ {train_dataset.__len__()} images)")
print(f"Val   batches per epoch: {len(val_loader)}   (≈ {val_dataset.__len__()} images)")
print(f"Test  batches per epoch: {len(test_loader)}  (≈ {test_dataset.__len__()} images)")
```

```python
# -------------------------------------------------
# Cell: Full training loop with mixed precision,
#       validation metrics, and checkpointing
# -------------------------------------------------

import torch
import torch.nn as nn
import torch.optim as optim
from torch.amp import GradScaler, autocast   # الاستيراد الحديث والموصى به
from torchvision import models

from sklearn.metrics import f1_score, accuracy_score
import numpy as np
import time
import copy
import os

# -------------------------------------------------------------------------
# 1️⃣ Device, model, loss, optimizer, scaler
# -------------------------------------------------------------------------
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"🖥️ Using device: {device}")

# Pretrained EfficientNet-B0 (torchvision) → 7-way classifier
model = models.efficientnet_b0(weights=models.EfficientNet_B0_Weights.DEFAULT)

# Replace the final fully-connected layer
in_features = model.classifier[1].in_features
model.classifier[1] = nn.Linear(in_features, 7)  # 7 disease classes

model = model.to(device)

# Loss – weighted CrossEntropy
criterion = nn.CrossEntropyLoss(weight=class_weights_torch.to(device))

# Optimiser – AdamW with weight-decay
optimizer = optim.AdamW(model.parameters(), lr=1e-3, weight_decay=1e-2)

# LR scheduler – ReduceLROnPlateau (monitor val-f1)
scheduler = optim.lr_scheduler.ReduceLROnPlateau(
    optimizer, mode='max', patience=2, factor=0.5
)

# Mixed-precision scaler (تحديد الجهاز تلقائياً وبأسلوب متوافق مع PyTorch الحديث)
scaler = GradScaler('cuda') if device.type == 'cuda' else None
print("🔀 Mixed-precision scaler:", scaler is not None)


# -------------------------------------------------------------------------
# 2️⃣ Helper: one training epoch
# -------------------------------------------------------------------------
def train_one_epoch(model, loader, optimizer, criterion, scaler):
    model.train()
    running_loss = 0.0
    all_labels = []
    all_preds  = []

    for batch_idx, (images, labels, *_) in enumerate(loader):
        images = images.to(device, non_blocking=True)
        labels = labels.to(device, non_blocking=True)

        optimizer.zero_grad()

        # Modern autocast (يحدد تفعيله تلقائياً بناءً على نوع الـ device)
        with autocast(device_type=device.type, enabled=(device.type == 'cuda')):
            outputs = model(images)
            loss = criterion(outputs, labels)

        # Backward pass (دعم الـ GPU مع Scaler أو الـ CPU العادي)
        if scaler is not None:
            scaler.scale(loss).backward()
            scaler.step(optimizer)
            scaler.update()
        else:
            loss.backward()
            optimizer.step()

        # Collect stats
        running_loss += loss.item() * images.size(0)

        _, preds = torch.max(outputs, 1)
        all_labels.extend(labels.cpu().numpy())
        all_preds.extend(preds.cpu().numpy())

    epoch_loss = running_loss / len(loader.dataset)
    epoch_acc  = accuracy_score(all_labels, all_preds)
    epoch_f1   = f1_score(all_labels, all_preds, average='macro', zero_division=0)

    return epoch_loss, epoch_acc, epoch_f1


# -------------------------------------------------------------------------
# 3️⃣ One validation epoch
# -------------------------------------------------------------------------
def validate_one_epoch(model, loader, criterion):
    model.eval()
    running_loss = 0.0
    all_labels = []
    all_preds  = []

    with torch.no_grad():
        for images, labels, *_ in loader:
            images = images.to(device, non_blocking=True)
            labels = labels.to(device, non_blocking=True)

            with autocast(device_type=device.type, enabled=(device.type == 'cuda')):
                outputs = model(images)
                loss = criterion(outputs, labels)

            running_loss += loss.item() * images.size(0)

            _, preds = torch.max(outputs, 1)
            all_labels.extend(labels.cpu().numpy())
            all_preds.extend(preds.cpu().numpy())

    val_loss = running_loss / len(loader.dataset)
    val_acc  = accuracy_score(all_labels, all_preds)
    val_f1   = f1_score(all_labels, all_preds, average='macro', zero_division=0)

    return val_loss, val_acc, val_f1


# -------------------------------------------------------------------------
# 4️⃣ Training loop with checkpointing and Early Stopping
# -------------------------------------------------------------------------
os.makedirs('data', exist_ok=True)  # التأكد من وجود مجلد حفظ النموذج

n_epochs = 15
best_f1  = -1.0
epochs_no_improve = 0

for epoch in range(1, n_epochs + 1):
    t0 = time.time()

    # ---- Train ----
    train_loss, train_acc, train_f1 = train_one_epoch(
        model, train_loader, optimizer, criterion, scaler
    )

    # ---- Validate ----
    val_loss, val_acc, val_f1 = validate_one_epoch(
        model, val_loader, criterion
    )

    # ---- Scheduler step ----
    scheduler.step(val_f1)

    # ---- Save best model ----
    if val_f1 > best_f1:
        best_f1 = val_f1
        torch.save(model.state_dict(), 'data/best_model.pth')
        epochs_no_improve = 0
        saved_str = f" | ✅ Saved Best (Val F1: {best_f1:.4f})"
    else:
        epochs_no_improve += 1
        saved_str = ""

    # ---- Clean CUDA cache ----
    if device.type == 'cuda':
        torch.cuda.empty_cache()

    # ---- Print epoch summary ----
    print(
        f"Epoch {epoch:02d}/{n_epochs} "
        f"| Train loss {train_loss:.4f} Acc {train_acc:.4f} F1 {train_f1:.4f} "
        f"| Val loss {val_loss:.4f} Acc {val_acc:.4f} F1 {val_f1:.4f} "
        f"| LR {optimizer.param_groups[0]['lr']:.1e} "
        f"| {time.time() - t0:.1f}s"
        f"{saved_str}"
    )

    # ---- Early stopping ----
    if epochs_no_improve >= 5:
        print("⏹️ Early stopping triggered.")
        break  # تم تصحيح مكان الـ break ليكون داخل شرط الـ if
```

```python
# -------------------------------------------------
# Cell: FocalLoss with Label Smoothing (PyTorch)
# -------------------------------------------------
import torch
import torch.nn as nn
import torch.nn.functional as F


class FocalLoss(nn.Module):
    """
    Focal Loss with optional Label Smoothing.

    Formula:
        FL = α_t * (1 - p_t)^γ * CE_smooth

    where:
        - p_t = predicted probability for the true class
        - CE_smooth = cross‑entropy with label‑smoothed targets
        - α_t = class‑weight for the true class (optional)
        - γ   = focusing parameter (default 2.0)

    Parameters
    ----------
    alpha : torch.Tensor or None, shape (C,)
        Class imbalance weights. If provided, α_t is looked up by class index.
    gamma : float
        Focusing parameter. Larger γ → down‑weight easy examples.
    label_smoothing : float
        Label smoothing factor ε ∈ [0, 1). 1 = uniform distribution.
    reduction : {'mean', 'sum', 'none'}
        How to aggregate the loss over the batch.
    """
    def __init__(self, alpha=None, gamma=2.0, label_smoothing=0.1, reduction='mean'):
        super(FocalLoss, self).__init__()
        self.alpha = alpha
        self.gamma = gamma
        self.label_smoothing = label_smoothing
        self.reduction = reduction

    def forward(self, inputs, targets):
        # -------------------------------------------------
        # inputs: (N, C) logits,  targets: (N) long class indices
        # -------------------------------------------------
        N, C = inputs.size()
        # 1️⃣ Log‑probabilities (numerically stable)
        log_ps = F.log_softmax(inputs, dim=1)          # (N, C)

        # 2️⃣ One‑hot encode targets
        one_hot = torch.zeros_like(inputs)
        one_hot.scatter_(1, targets.unsqueeze(1), 1)   # (N, C), 1 at target pos

        # 3️⃣ Apply label smoothing to the one‑hot targets
        #    → smoothed_target = (1 - ε) * one_hot + ε / C
        eps = self.label_smoothing
        targets_smooth = one_hot * (1.0 - eps) + (eps / C)

        # 4️⃣ Cross‑Entropy loss with smoothed targets (per sample)
        #    CE = - Σ_j smoothed_target_j * log_ps_j
        ce_loss = -(targets_smooth * log_ps).sum(dim=1)   # (N,)

        # 5️⃣ Compute pt = predicted probability of the true class
        ps = torch.softmax(inputs, dim=1)                # (N, C)
        pt = ps.gather(1, targets.unsqueeze(1)).squeeze(1)   # (N,)

        # 6️⃣ Modulating factor (1 - pt)^γ
        modulating_factor = (1.0 - pt) ** self.gamma

        # 7️⃣ Alpha (class‑weight) weighting
        if self.alpha is not None:
            # ensure alpha is on the same device as inputs
            alpha = self.alpha.to(inputs.device)
            # gather the weight for each sample's target class
            at = alpha.gather(0, targets)          # (N,)
        else:
            at = torch.ones_like(ce_loss)

        # 8️⃣ Final focal loss per sample
        loss = at * modulating_factor * ce_loss

        # 9️⃣ Reduction
        if self.reduction == 'mean':
            return loss.mean()
        elif self.reduction == 'sum':
            return loss.sum()
        else:  # 'none'
            return loss


# -------------------------------------------------
# Instantiation with our previously computed class weights
# -------------------------------------------------
# class_weights_torch should already be on the correct device,
# shape = (num_classes,) = (7,) for HAM10000
criterion = FocalLoss(
    alpha=class_weights_torch.to(device),   # ← passes the balanced weights
    gamma=2.0,
    label_smoothing=0.1,
    reduction='mean'
)

print("✅ FocalLoss initialized with gamma=2.0 and label_smoothing=0.1")
```

```python
# -------------------------------------------------
# Cell: Evaluate clinical‑best model (EfficientNet‑B3) with TTA
# -------------------------------------------------
import torch
import torch.nn.functional as F
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import torchvision.models as models
from torchvision.models.efficientnet import EfficientNet_B3_Weights
from sklearn.metrics import classification_report, confusion_matrix


# -------------------------------------------------
# 1️⃣ Build the same EfficientNet‑B3 architecture used during training
# -------------------------------------------------
model = models.efficientnet_b3(weights=EfficientNet_B3_Weights.DEFAULT)

# Replace the final classifier to output 7 disease classes
in_features = model.classifier[1].in_features          # 1536 for B3
model.classifier = torch.nn.Sequential(
    torch.nn.Dropout(p=0.4, inplace=False),
    torch.nn.Linear(in_features, 7)                     # 7 classes
)

# Load the saved checkpoint
checkpoint = torch.load('data/clinical_best_model.pth', map_location=device)
model.load_state_dict(checkpoint)
model = model.to(device)
model.eval()                                       # disables dropout / BN updates
print("✅ Model loaded and set to eval mode")


# -------------------------------------------------
# 2️⃣ Helper: apply a flip to a batch and return logits
# -------------------------------------------------
def apply_flip(images, flip_type):
    """
    Flip a batch of images along the specified axis.
    flip_type: 'horizontal' → flip width (dim = 3), 'vertical' → flip height (dim = 2)
    """
    if flip_type == 'horizontal':
        return torch.flip(images, dims=[3])
    elif flip_type == 'vertical':
        return torch.flip(images, dims=[2])
    else:
        raise ValueError("flip_type must be 'horizontal' or 'vertical'")


# -------------------------------------------------
# 3️⃣ Inference loop with 3‑fold TTA
# -------------------------------------------------
all_true = []      # ground‑truth class indices (0‑6)
all_pred = []      # 🟢 تم إضافة تعريف القائمة هنا لمنع الخطأ
all_prob = []      # averaged soft‑probabilities (7‑dim per sample)

with torch.no_grad():
    for images, labels, _ in test_loader:
        images = images.to(device, non_blocking=True)
        labels = labels.to(device, non_blocking=True)

        # ---- Original image ----
        with torch.amp.autocast('cuda' if torch.cuda.is_available() else 'cpu'):
            logits0 = model(images)
        prob0 = torch.softmax(logits0, dim=1)

        # ---- Horizontal flip ----
        flipped_h = apply_flip(images, 'horizontal')
        with torch.amp.autocast('cuda' if torch.cuda.is_available() else 'cpu'):
            logits_h = model(flipped_h)
        prob_h = torch.softmax(logits_h, dim=1)

        # ---- Vertical flip ----
        flipped_v = apply_flip(images, 'vertical')
        with torch.amp.autocast('cuda' if torch.cuda.is_available() else 'cpu'):
            logits_v = model(flipped_v)
        prob_v = torch.softmax(logits_v, dim=1)

        # Average the three probability maps (original + two flips)
        avg_prob = (prob0 + prob_h + prob_v) / 3.0

        # Collect predictions (argmax) and probabilities
        pred = torch.argmax(avg_prob, dim=1).cpu().numpy()
        true = labels.cpu().numpy()

        all_pred.extend(pred.tolist())
        all_true.extend(true.tolist())
        all_prob.extend(avg_prob.cpu().numpy())

all_true = np.array(all_true)
all_pred = np.array(all_pred)
all_prob = np.array(all_prob)

print(f"🔢 Collected {len(all_true)} predictions from {len(test_loader)} test batches")


# -------------------------------------------------
# 4️⃣ Classification report (human‑readable names, 4‑digit precision)
# -------------------------------------------------
class_names = [idx_to_dx[i] for i in sorted(idx_to_dx.keys())]

print("\n=== Classification Report (precision, recall, f1‑score) ===")
report = classification_report(
    all_true,
    all_pred,
    labels=sorted(idx_to_dx.keys()),
    target_names=class_names,
    digits=4,
    zero_division=0
)
print(report)


# -------------------------------------------------
# 5️⃣ Confusion matrix heatmap (raw integer counts)
# -------------------------------------------------
cm = confusion_matrix(all_true, all_pred, labels=sorted(idx_to_dx.keys()))

plt.figure(figsize=(9, 7))
sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    cmap='Blues',
    xticklabels=class_names,
    yticklabels=class_names,
    cbar=False
)
plt.title('Confusion Matrix – Test Set (EfficientNet‑B3, Phase 3 TTA)', fontsize=14)
plt.ylabel('True Diagnosis', fontsize=12)
plt.xlabel('Predicted Diagnosis', fontsize=12)
plt.tight_layout()
plt.show()
```

```python
# -------------------------------------------------
# Cell: Grid‑search melanoma threshold
# -------------------------------------------------
import numpy as np
import pandas as pd
from sklearn.metrics import classification_report, recall_score, f1_score

# -------------------------------------------------
# 1️⃣  Helper: turn probabilities into predictions
# -------------------------------------------------
def apply_mel_threshold(y_probs, mel_thr, mel_idx, y_true):
    """
    For each sample:
        - if prob[mel] >= mel_thr → predict melanoma (mel_idx)
        - else → predict the class with the highest probability
          (other classes keep their usual arg‑max behaviour).
    Returns an array of predicted class indices (same length as y_true).
    """
    y_pred = np.empty(y_probs.shape[0], dtype=int)
    for i, probs in enumerate(y_probs):
        if probs[mel_idx] >= mel_thr:
            y_pred[i] = mel_idx
        else:
            y_pred[i] = np.argmax(probs)        # default arg‑max rule
    return y_pred


# -------------------------------------------------
# 2️⃣  Prepare data
# -------------------------------------------------
# locate the index for the melanoma class (mel)
mel_idx = next(k for k, v in idx_to_dx.items() if v == "mel")

# all candidate thresholds: 0.25, 0.30, 0.35, 0.40, 0.45
thresholds = np.arange(0.25, 0.455, 0.05)

# -------------------------------------------------
# 3️⃣  Grid‑search
# -------------------------------------------------
best_thr = None
best_f1  = -1.0
best_rec = -1.0
results = []          # store (threshold, f1, recall) for inspection

# Getting sorted label indices
labels = sorted(idx_to_dx.keys())

for thr in thresholds:
    y_pred = apply_mel_threshold(y_probs, thr, mel_idx, y_true)

    # Calculate per-class recall and F1 using average=None
    rec_per_class = recall_score(y_true, y_pred, labels=labels, average=None, zero_division=0)
    f1_per_class  = f1_score(y_true, y_pred, labels=labels, average=None, zero_division=0)

    # Extract metrics for melanoma (using its positional index in sorted labels)
    mel_pos = labels.index(mel_idx)
    rec = rec_per_class[mel_pos]
    f1  = f1_per_class[mel_pos]

    results.append((thr, f1, rec))

    # Constraint: Recall >= 0.70 while maximizing F1
    if rec >= 0.70:
        if f1 > best_f1:
            best_f1  = f1
            best_rec = rec
            best_thr = thr

# -------------------------------------------------
# 4️⃣  Output
# -------------------------------------------------
print("=== Grid‑search results (threshold, melanoma F1, melanoma Recall) ===")
for t, f, r in results:
    print(f"threshold={t:.2f}  →  F1={f:.4f}   Recall={r:.4f}")

if best_thr is not None:
    print(f"\n✅ Best threshold found: {best_thr:.2f}")
    print(f"   – Melanoma Recall = {best_rec:.2%}  (≥ 70% required)")
    print(f"   – Melanoma F1‑Score = {best_f1:.4f}")

    # Build the final predictions with the best threshold
    y_pred_best = apply_mel_threshold(y_probs, best_thr, mel_idx, y_true)

    # Print the full classification report
    target_names = [idx_to_dx[i] for i in labels]
    print("\n=== Fine‑tuned Classification Report (mel threshold {:.2f}) ===".format(best_thr))
    print(classification_report(
        y_true,
        y_pred_best,
        labels=labels,
        target_names=target_names,
        digits=4,
        zero_division=0,
    ))
else:
    print("\n⚠️  No threshold satisfied the recall ≥ 70% constraint; "
          "consider widening the search range or adjusting the constraint.")
```

```python
# -------------------------------------------------
# Cell: ConvNeXt‑Base clinical‑grade training + TTA eval
# -------------------------------------------------
# ► 1️⃣   Imports, GPU Purge & Reproducibility
# -------------------------------------------------
import os, random, time, math, gc
from pathlib import Path
from collections import Counter

import numpy as np
import pandas as pd
from PIL import Image
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader, WeightedRandomSampler
from torchvision import models
from torchvision.models.convnext import ConvNeXt_Base_Weights

import albumentations as A
from albumentations.pytorch import ToTensorV2

from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score, recall_score, f1_score
import matplotlib.pyplot as plt
import seaborn as sns

# Force garbage collection and flush CUDA cache completely
gc.collect()
if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.ipc_collect()

# Enable PyTorch memory fragmentation allocator optimization
os.environ["PYTORCH_CUDA_ALLOC_CONF"] = "expandable_segments:True"

# -------------------------------------------------
# ► 2️⃣   Hyper‑parameters & seed
# -------------------------------------------------
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"🔧 Device: {device}")

# 4GB VRAM Optimization Parameters:
# Downscaling resolution to 256x256 drops activation memory by ~55%
IMG_SIZE = 256
BATCH_SIZE = 4               # Physical batch per step
ACCUMULATION_STEPS = 4        # Effective batch size = 16 (4 * 4)
NUM_EPOCHS = 15
LR_BACKBONE = 1e-5           # ConvNeXt base LR
LR_HEAD     = 1e-4           # Classifier LR
WEIGHT_DECAY = 1e-2
PATIENCE_RLR = 2            # For ReduceLROnPlateau
FLT_LABEL_SMOOTHING = 0.05
NUM_WORKERS = 0              # Prevents deadlock on Windows multiprocessing

# -------------------------------------------------
# ► 3️⃣   Albumentations transforms (256×256)
# -------------------------------------------------
def get_train_transform():
    return A.Compose([
        A.Resize(IMG_SIZE, IMG_SIZE),
        A.HorizontalFlip(p=0.5),
        A.VerticalFlip(p=0.5),
        A.ShiftScaleRotate(shift_limit=0.1, scale_limit=0.1, rotate_limit=30, p=0.5),
        A.ColorJitter(brightness=0.2, contrast=0.2, saturation=0.2, hue=0.1, p=0.4),
        A.Normalize(mean=[0.485, 0.456, 0.406],
                    std=[0.229, 0.224, 0.225],
                    max_pixel_value=255.0),
        ToTensorV2(),
    ])

def get_val_test_transform():
    return A.Compose([
        A.Resize(IMG_SIZE, IMG_SIZE),
        A.Normalize(mean=[0.485, 0.456, 0.406],
                    std=[0.229, 0.224, 0.225],
                    max_pixel_value=255.0),
        ToTensorV2(),
    ])

# -------------------------------------------------
# ► 4️⃣   Dataset loader
# -------------------------------------------------
class Ham10000Dataset(Dataset):
    def __init__(self, df, transform, image_folders):
        self.df = df.reset_index(drop=True)
        self.transform = transform
        self.idx2path = {}
        for folder in image_folders:
            if not os.path.exists(folder):
                continue
            for fn in os.listdir(folder):
                if fn.lower().endswith(('.png', '.jpg', '.jpeg')):
                    img_id = fn.rsplit('.', 1)[0]
                    self.idx2path[img_id] = os.path.join(folder, fn)

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        row = self.df.iloc[idx]
        img_id = row['image_id']
        dx_code = row['dx']

        if img_id not in self.idx2path:
            raise FileNotFoundError(f"Image {img_id} not found on disk")

        img_path = self.idx2path[img_id]
        image = Image.open(img_path).convert('RGB')
        image = np.array(image)

        if self.transform:
            augmented = self.transform(image=image)
            image = augmented['image']

        label = dx_to_idx[dx_code]
        return image, label, img_id


# -------------------------------------------------
# ► 5️⃣   Load metadata & create train/val/test splits
# -------------------------------------------------
metadata_path = "data/HAM10000_metadata.csv"
df_all = pd.read_csv(metadata_path)

splits_dir = Path("data")
splits_dir.mkdir(parents=True, exist_ok=True)
split_files = [splits_dir/"train_metadata.csv",
               splits_dir/"val_metadata.csv",
               splits_dir/"test_metadata.csv"]

if not all(f.exists() for f in split_files):
    print("⚙️  Pre‑made splits not found – creating 80/10/10 stratified split")
    X = df_all['image_id']
    y = df_all['dx']
    skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=SEED)
    for tr_idx, te_idx in skf.split(X, y):
        df_train_val = df_all.iloc[tr_idx]
        df_test      = df_all.iloc[te_idx]
        break
    skf2 = StratifiedKFold(n_splits=2, shuffle=True, random_state=SEED)
    for tr_idx, va_idx in skf2.split(df_train_val['image_id'], df_train_val['dx']):
        df_train = df_train_val.iloc[tr_idx]
        df_val   = df_train_val.iloc[va_idx]
        break

    df_train.to_csv(splits_dir/"train_metadata.csv", index=False)
    df_val.to_csv(splits_dir/"val_metadata.csv",      index=False)
    df_test.to_csv(splits_dir/"test_metadata.csv",   index=False)
    print("✅ Splits created and saved.")
else:
    df_train = pd.read_csv(splits_dir/"train_metadata.csv")
    df_val   = pd.read_csv(splits_dir/"val_metadata.csv")
    df_test  = pd.read_csv(splits_dir/"test_metadata.csv")
    print("✅ Using existing splits.")

# -------------------------------------------------
# ► 6️⃣   Label mapping
# -------------------------------------------------
dx_codes = sorted(df_all['dx'].unique())
dx_to_idx = {code: i for i, code in enumerate(dx_codes)}
idx_to_dx = {v: k for k, v in dx_to_idx.items()}
NUM_CLASSES = len(dx_codes)
print(f"🗺️  Label map: {dx_to_idx}")

# -------------------------------------------------
# ► 7️⃣   Build Datasets & WeightedRandomSampler
# -------------------------------------------------
IMAGE_FOLDERS = ["data/HAM10000_images_part_1",
                 "data/HAM10000_images_part_2"]

train_ds = Ham10000Dataset(df=df_train, transform=get_train_transform(),
                           image_folders=IMAGE_FOLDERS)
val_ds   = Ham10000Dataset(df=df_val,   transform=get_val_test_transform(),
                           image_folders=IMAGE_FOLDERS)
test_ds  = Ham10000Dataset(df=df_test,  transform=get_val_test_transform(),
                           image_folders=IMAGE_FOLDERS)

class_counts = train_ds.df['dx'].value_counts()
inv_freq = {cls: 1.0 / count for cls, count in class_counts.items()}
sample_weights = train_ds.df['dx'].map(inv_freq).values
sample_weight_tensor = torch.tensor(sample_weights, dtype=torch.float32)

sampler = WeightedRandomSampler(
    weights=sample_weight_tensor,
    num_samples=len(sample_weight_tensor),
    replacement=True
)

train_loader = DataLoader(train_ds, batch_size=BATCH_SIZE, sampler=sampler, num_workers=NUM_WORKERS, pin_memory=True)
val_loader   = DataLoader(val_ds,   batch_size=BATCH_SIZE, shuffle=False, num_workers=NUM_WORKERS, pin_memory=True)
test_loader  = DataLoader(test_ds,  batch_size=BATCH_SIZE, shuffle=False, num_workers=NUM_WORKERS, pin_memory=True)

print(f"⚖️  WeightedRandomSampler created – {len(sample_weight_tensor)} training samples")

# -------------------------------------------------
# ► 8️⃣   Model: ConvNeXt‑Base + custom head
# -------------------------------------------------
model = models.convnext_base(weights=ConvNeXt_Base_Weights.DEFAULT)

in_features = model.classifier[2].in_features

model.classifier = nn.Sequential(
    nn.LayerNorm((in_features, 1, 1)),
    nn.Flatten(start_dim=1),
    nn.Dropout(p=0.4, inplace=False),
    nn.Linear(in_features, NUM_CLASSES)
)

# Allocate to device safely
model = model.to(device)
print(f"🏗️  ConvNeXt‑Base created, head → LayerNorm → Flatten → Dropout(0.4) → Linear({in_features} → {NUM_CLASSES})")

# -------------------------------------------------
# ► 9️⃣   Loss, optimizer, scheduler
# -------------------------------------------------
criterion = nn.CrossEntropyLoss(label_smoothing=FLT_LABEL_SMOOTHING).to(device)

optimizer = optim.AdamW([
    {"params": model.features.parameters(), "lr": LR_BACKBONE},
    {"params": model.classifier.parameters(), "lr": LR_HEAD}
], weight_decay=WEIGHT_DECAY)

scheduler = optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode='max', factor=0.5, patience=PATIENCE_RLR)

scaler = torch.amp.GradScaler('cuda') if torch.cuda.is_available() else None
print("🔀  Mixed‑precision enabled" if scaler else "⚠️  Mixed‑precision disabled")

# -------------------------------------------------
# ►10️⃣   Helper functions – train / validation
# -------------------------------------------------
def train_one_epoch(model, loader, optimizer, criterion, scaler):
    model.train()
    running_loss = 0.0
    all_labels, all_preds = [], []

    optimizer.zero_grad()
    for step, (images, labels, _) in enumerate(loader):
        images = images.to(device, non_blocking=True)
        labels = labels.to(device, non_blocking=True)

        if scaler is not None:
            with torch.autocast(device_type='cuda'):
                logits = model(images)
                loss = criterion(logits, labels)
                loss = loss / ACCUMULATION_STEPS

            scaler.scale(loss).backward()
            
            if (step + 1) % ACCUMULATION_STEPS == 0 or (step + 1) == len(loader):
                scaler.step(optimizer)
                scaler.update()
                optimizer.zero_grad()
        else:
            logits = model(images)
            loss = criterion(logits, labels)
            loss = loss / ACCUMULATION_STEPS
            loss.backward()

            if (step + 1) % ACCUMULATION_STEPS == 0 or (step + 1) == len(loader):
                optimizer.step()
                optimizer.zero_grad()

        running_loss += (loss.item() * ACCUMULATION_STEPS) * images.size(0)

        with torch.no_grad():
            preds = torch.argmax(logits, dim=1).cpu().numpy()
        all_labels.extend(labels.cpu().numpy())
        all_preds.extend(preds)

    epoch_loss = running_loss / len(loader.dataset)
    acc = accuracy_score(all_labels, all_preds)
    macro_f1 = f1_score(all_labels, all_preds, average='macro', zero_division=0)
    macro_rec = recall_score(all_labels, all_preds, average='macro', zero_division=0)
    return epoch_loss, acc, macro_f1, macro_rec, all_labels, all_preds


def validate_one_epoch(model, loader, criterion):
    model.eval()
    running_loss = 0.0
    all_labels, all_preds = [], []

    with torch.no_grad():
        for images, labels, _ in loader:
            images = images.to(device, non_blocking=True)
            labels = labels.to(device, non_blocking=True)

            if torch.cuda.is_available():
                with torch.autocast(device_type='cuda'):
                    logits = model(images)
                    loss = criterion(logits, labels)
            else:
                logits = model(images)
                loss = criterion(logits, labels)

            running_loss += loss.item() * images.size(0)

            preds = torch.argmax(logits, dim=1).cpu().numpy()
            all_labels.extend(labels.cpu().numpy())
            all_preds.extend(preds)

    epoch_loss = running_loss / len(loader.dataset)
    acc = accuracy_score(all_labels, all_preds)
    macro_f1 = f1_score(all_labels, all_preds, average='macro', zero_division=0)
    macro_rec = recall_score(all_labels, all_preds, average='macro', zero_division=0)
    return epoch_loss, acc, macro_f1, macro_rec, all_labels, all_preds


# -------------------------------------------------
# ►11️⃣   15‑epoch training loop
# -------------------------------------------------
BEST_VAL_F1 = -1.0
CHECKPOINT_PATH = "data/convnext_clinical_best.pth"

for epoch in range(1, NUM_EPOCHS + 1):
    t0 = time.time()

    train_loss, train_acc, train_f1, train_rec, _, _ = train_one_epoch(
        model, train_loader, optimizer, criterion, scaler)

    val_loss, val_acc, val_f1, val_rec, _, _ = validate_one_epoch(
        model, val_loader, criterion)

    scheduler.step(val_f1)

    if val_f1 > BEST_VAL_F1:
        BEST_VAL_F1 = val_f1
        torch.save(model.state_dict(), CHECKPOINT_PATH)
        print(f"✅ Epoch {epoch:02d}: new best val‑Macro‑F1 {BEST_VAL_F1:.4f} → saved {CHECKPOINT_PATH}")

    print(
        f"Epoch {epoch:02d}/{NUM_EPOCHS} | "
        f"Train L {train_loss:.4f} Acc {train_acc:.4f} F1 {train_f1:.4f} | "
        f"Val L {val_loss:.4f} Acc {val_acc:.4f} F1 {val_f1:.4f} | "
        f"LR {optimizer.param_groups[0]['lr']:.1e} | "
        f"{time.time() - t0:.1f}s"
    )

    if torch.cuda.is_available():
        torch.cuda.empty_cache()

print("\n✅ Training finished!")

# -------------------------------------------------
# ►12️⃣   Test‑Time Augmentation (TTA) evaluation
# -------------------------------------------------
def tta_predict(model, loader):
    model.eval()
    all_probs, all_labels = [], []

    with torch.no_grad():
        for images, labels, _ in loader:
            images = images.to(device, non_blocking=True)
            labels = labels.to(device, non_blocking=True)

            if torch.cuda.is_available():
                with torch.autocast(device_type='cuda'):
                    logits0 = model(images)
                probs0 = torch.softmax(logits0, dim=1).cpu().numpy()

                flipped_h = torch.flip(images, dims=[3])
                with torch.autocast(device_type='cuda'):
                    logits_h = model(flipped_h)
                probs_h = torch.softmax(logits_h, dim=1).cpu().numpy()

                flipped_v = torch.flip(images, dims=[2])
                with torch.autocast(device_type='cuda'):
                    logits_v = model(flipped_v)
                probs_v = torch.softmax(logits_v, dim=1).cpu().numpy()
            else:
                logits0 = model(images)
                probs0 = torch.softmax(logits0, dim=1).cpu().numpy()

                flipped_h = torch.flip(images, dims=[3])
                logits_h = model(flipped_h)
                probs_h = torch.softmax(logits_h, dim=1).cpu().numpy()

                flipped_v = torch.flip(images, dims=[2])
                logits_v = model(flipped_v)
                probs_v = torch.softmax(logits_v, dim=1).cpu().numpy()

            avg_probs = (probs0 + probs_h + probs_v) / 3.0
            all_probs.extend(avg_probs)
            all_labels.extend(labels.cpu().numpy())

    return np.array(all_probs), np.array(all_labels)

if os.path.exists(CHECKPOINT_PATH):
    model.load_state_dict(torch.load(CHECKPOINT_PATH, map_location=device))
    print(f"📂 Loaded best model from {CHECKPOINT_PATH}")

tta_probs, tta_labels = tta_predict(model, test_loader)
tta_preds = np.argmax(tta_probs, axis=1)
target_names = [idx_to_dx[i] for i in sorted(idx_to_dx.keys())]

print("\n=== TTA‑Enhanced Classification Report (4 d.p.) ===")
print(classification_report(tta_labels, tta_preds, target_names=target_names, digits=4, zero_division=0))

cm = confusion_matrix(tta_labels, tta_preds, labels=sorted(idx_to_dx.keys()))
plt.figure(figsize=(9, 7))
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", xticklabels=target_names, yticklabels=target_names, cbar=False)
plt.title("TTA‑Enhanced Confusion Matrix")
plt.ylabel("True Diagnosis")
plt.xlabel("Predicted Diagnosis")
plt.tight_layout()
plt.show()
```

```python
# -------------------------------------------------
# Cell: Grad-CAM Visualization for ConvNeXt-Base
# -------------------------------------------------
import cv2
import torch
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

# 1. Grad-CAM Class Definition
class ConvNeXtGradCAM:
    def __init__(self, model, target_layer):
        self.model = model
        self.target_layer = target_layer
        self.gradients = None
        self.activations = None
        
        # Hooks to capture activations and gradients
        self.target_layer.register_forward_hook(self._save_activation)
        self.target_layer.register_full_backward_hook(self._save_gradient)

    def _save_activation(self, module, input, output):
        self.activations = output

    def _save_gradient(self, module, grad_input, grad_output):
        self.gradients = grad_output[0]

    def generate(self, input_tensor, target_class=None):
        self.model.eval()
        output = self.model(input_tensor)

        if target_class is None:
            target_class = torch.argmax(output, dim=1).item()

        self.model.zero_grad()
        score = output[0, target_class]
        score.backward()

        # Channel-wise intensity weighting
        gradients = self.gradients.data.cpu().numpy()[0]
        activations = self.activations.data.cpu().numpy()[0]
        weights = np.mean(gradients, axis=(1, 2))

        cam = np.zeros(activations.shape[1:], dtype=np.float32)
        for i, w in enumerate(weights):
            cam += w * activations[i]

        cam = np.maximum(cam, 0)
        if cam.max() > 0:
            cam = cam / cam.max()
        return cam, target_class

# 2. Helper function to plot image and heatmap
def plot_gradcam(image_path, model, transform):
    raw_img = Image.open(image_path).convert('RGB')
    raw_img_resized = raw_img.resize((IMG_SIZE, IMG_SIZE))
    
    # Process image
    img_array = np.array(raw_img)
    augmented = transform(image=img_array)
    input_tensor = augmented['image'].unsqueeze(0).to(device)

    # Initialize CAM on ConvNeXt last feature layer
    target_layer = model.features[-1]
    grad_cam = ConvNeXtGradCAM(model, target_layer)

    cam, pred_class = grad_cam.generate(input_tensor)

    # Resize CAM heatmap to match input size
    heatmap = cv2.resize(cam, (IMG_SIZE, IMG_SIZE))
    heatmap = np.uint8(255 * heatmap)
    heatmap = cv2.applyColorMap(heatmap, cv2.COLORMAP_JET)
    heatmap = cv2.cvtColor(heatmap, cv2.COLOR_BGR2RGB)

    # Overlay heatmap on original image
    img_np = np.array(raw_img_resized)
    superimposed_img = cv2.addWeighted(img_np, 0.6, heatmap, 0.4, 0)

    # Visualization
    fig, axes = plt.subplots(1, 3, figsize=(15, 5))
    axes[0].imshow(img_np)
    axes[0].set_title("Original Image")
    axes[0].axis('off')

    axes[1].imshow(heatmap)
    axes[1].set_title("Heatmap (Activation)")
    axes[1].axis('off')

    axes[2].imshow(superimposed_img)
    axes[2].set_title(f"Grad-CAM (Pred: {CLASS_NAMES[pred_class]})")
    axes[2].axis('off')

    plt.tight_layout()
    plt.show()

# -------------------------------------------------
# 3. Test on a Sample Image from Test Metadata
# -------------------------------------------------
sample_row = df_test.iloc[0]  # اختر أي صورة (مثلاً السطر الأول)
sample_id = sample_row['image_id']
sample_dx = sample_row['dx']

# Locate image path
sample_path = test_dataset.idx2path.get(sample_id)

print(f"🔬 Testing Grad-CAM for Image: {sample_id} | True Label: {sample_dx}")
plot_gradcam(sample_path, model, get_test_transform())
```

```python
import os
import argparse
import torch
import torch.nn as nn
import numpy as np
from PIL import Image
import albumentations as A
from albumentations.pytorch import ToTensorV2
from torchvision import models

# -------------------------------------------------
# 1. الإعدادات والمعلمات الثابتة
# -------------------------------------------------
IMG_SIZE = 384
NUM_CLASSES = 7
MEL_IDX = 4
MEL_THRESHOLD = 0.30  # العتبة المحسنة للميلانوما

CLASS_NAMES = ["akiec", "bcc", "bkl", "df", "mel", "nv", "vasc"]
CLASS_DESCRIPTIONS = {
    "akiec": "Actinic keratoses / Bowen's disease (Pre-cancerous)",
    "bcc": "Basal cell carcinoma (Malignant)",
    "bkl": "Benign keratosis-like lesions (Benign)",
    "df": "Dermatofibroma (Benign)",
    "mel": "Melanoma (Malignant - High Risk)",
    "nv": "Melanocytic nevi (Benign mole)",
    "vasc": "Vascular lesions (Benign)"
}

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# -------------------------------------------------
# 2. بناء الهيكل الهندسي للموديل وتحميل الأوزان
# -------------------------------------------------
class LayerNorm2d(nn.Module):
    def __init__(self, channels, eps=1e-6):
        super().__init__()
        self.weight = nn.Parameter(torch.ones(channels, 1, 1))
        self.bias = nn.Parameter(torch.zeros(channels, 1, 1))
        self.eps = eps

    def forward(self, x):
        u = x.mean(1, keepdim=True)
        s = (x - u).pow(2).mean(1, keepdim=True)
        x = (x - u) / torch.sqrt(s + self.eps)
        return self.weight * x + self.bias

def load_clinical_model(ckpt_path="data/convnext_clinical_best.pth"):
    model = models.convnext_base(weights=None)
    in_features = model.classifier[2].in_features
    
    model.classifier = nn.Sequential(
        LayerNorm2d(in_features),
        nn.Flatten(start_dim=1),
        nn.Dropout(p=0.4, inplace=False),
        nn.Linear(in_features, NUM_CLASSES)
    )
    
    if not os.path.exists(ckpt_path):
        raise FileNotFoundError(f"لم يتم العثور على ملف الأوزان في: {ckpt_path}")
        
    ckpt = torch.load(ckpt_path, map_location=device)
    state_dict = ckpt["state_dict"] if isinstance(ckpt, dict) and "state_dict" in ckpt else (ckpt["model"] if isinstance(ckpt, dict) and "model" in ckpt else ckpt)
    state_dict = {k.replace("module.", ""): v for k, v in state_dict.items()}
    
    model.load_state_dict(state_dict)
    model = model.to(device)
    model.eval()
    return model

# -------------------------------------------------
# 3. المعالجة الأولية للصورة (Transformations)
# -------------------------------------------------
def transform_image(image_path):
    transform = A.Compose([
        A.Resize(IMG_SIZE, IMG_SIZE, always_apply=True),
        A.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225], max_pixel_value=255.0),
        ToTensorV2(),
    ])
    
    raw_img = Image.open(image_path).convert('RGB')
    img_np = np.array(raw_img)
    augmented = transform(image=img_np)
    tensor = augmented['image'].unsqueeze(0).to(device)
    return tensor

# -------------------------------------------------
# 4. دالة التنبؤ (Inference Engine)
# -------------------------------------------------
def predict_skin_lesion(image_path, model, apply_threshold=True):
    input_tensor = transform_image(image_path)
    
    with torch.no_grad():
        logits = model(input_tensor)
        probs = torch.softmax(logits, dim=1).cpu().numpy()[0]
        
    mel_prob = probs[MEL_IDX]
    
    # تطبيق منطق العتبة المحسنة (Optimized Thresholding)
    if apply_threshold and mel_prob >= MEL_THRESHOLD:
        pred_idx = MEL_IDX
    else:
        # اختيار الفئة الأعلى احتمالاً من بين البقية لو لم تتخطى الميلانوما العتبة
        temp_probs = probs.copy()
        if apply_threshold:
            temp_probs[MEL_IDX] = -1.0
        pred_idx = np.argmax(temp_probs)
        
    pred_label = CLASS_NAMES[pred_idx]
    confidence = probs[pred_idx] * 100

    return {
        "pred_label": pred_label,
        "description": CLASS_DESCRIPTIONS[pred_label],
        "confidence": confidence,
        "mel_probability": mel_prob * 100,
        "all_probabilities": {CLASS_NAMES[i]: float(probs[i] * 100) for i in range(NUM_CLASSES)}
    }

# -------------------------------------------------
# 5. واجهة التشغيل المباشرة
# -------------------------------------------------
if __name__ == "__main__":
    # مسار الصورة للاختبار (استبدله بمسار أي صورة عندك)
    sample_image_path = "D:\images.jpg"
    
    print("⏳ جاري تحميل الموديل...")
    clinical_model = load_clinical_model()
    
    if os.path.exists(sample_image_path):
        print(f"🔬 جاري تحليل الصورة: {sample_image_path}\n")
        res = predict_skin_lesion(sample_image_path, clinical_model)
        
        print("=" * 55)
        print(f" 🎯 التشخيص النهائي: {res['pred_label'].upper()}")
        print(f" 📝 الوصف الطبي   : {res['description']}")
        print(f" 🛡️  نسبة الثقة    : {res['confidence']:.2f}%")
        print(f" ⚠️  احتمال Melanoma: {res['mel_probability']:.2f}% (Threshold = {MEL_THRESHOLD*100}%)")
        print("=" * 55)
        
        print("\n📊 توزيع الاحتمالات لكافة الفئات:")
        for cls_name, prob in res["all_probabilities"].items():
            bar = "█" * int(prob / 5)
            print(f"  - {cls_name:5s}: {prob:6.2f}% | {bar}")
    else:
        print(f"❌ لم يتم العثور على الصورة في المسار المحدد: {sample_image_path}")
```
